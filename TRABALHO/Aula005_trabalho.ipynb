Dataset (base do problema)

```python
import pandas as pd
import numpy as np
from math import log2

data = [
    ["jovem", "miopia", "não", "reduzida", "nenhuma"],
    ["jovem", "miopia", "sim", "normal", "gelatinosa"],
    ["jovem", "hipermetropia", "não", "normal", "gelatinosa"],
    ["jovem", "hipermetropia", "sim", "normal", "dura"],
    ["pré-presbiopia", "miopia", "não", "reduzida", "gelatinosa"],
    ["pré-presbiopia", "miopia", "sim", "reduzida", "nenhuma"],
    ["pré-presbiopia", "miopia", "não", "normal", "dura"],
    ["pré-presbiopia", "hipermetropia", "não", "reduzida", "gelatinosa"],
    ["pré-presbiopia", "hipermetropia", "sim", "normal", "dura"],
    ["presbiopia", "miopia", "não", "normal", "gelatinosa"],
    ["presbiopia", "miopia", "sim", "normal", "dura"],
    ["presbiopia", "miopia", "sim", "normal", "gelatinosa"],
    ["presbiopia", "hipermetropia", "não", "reduzida", "nenhuma"],
    ["presbiopia", "hipermetropia", "sim", "normal", "gelatinosa"],
    ["presbiopia", "hipermetropia", "não", "normal", "gelatinosa"]
]

columns = ["Idade", "Diagnóstico", "Astigmatismo", "Taxa_lacrimal", "Tipo_de_Lente"]

df = pd.DataFrame(data, columns=columns)

df
```

Função de Entropia


```python
def entropia(coluna):
    valores, contagens = np.unique(coluna, return_counts=True)
    probs = contagens / contagens.sum()
    return -np.sum(probs * np.log2(probs))
```

Entropia Total (igual início da resolução)

```python
ent_total = entropia(df["Tipo_de_Lente"])
print("Entropia Total:", round(ent_total, 2))
```

Função de Ganho de Informação

```python
def ganho_informacao(df, atributo, alvo="Tipo_de_Lente"):
    entropia_total = entropia(df[alvo])

    entropia_condicional = 0

    for valor in df[atributo].unique():
        subset = df[df[atributo] == valor]
        peso = len(subset) / len(df)
        entropia_subset = entropia(subset[alvo])

        print(f"{atributo} = {valor}")
        print(f"Peso: {round(peso,2)} | Entropia: {round(entropia_subset,2)}\n")

        entropia_condicional += peso * entropia_subset

    ganho = entropia_total - entropia_condicional

    print("Ganho:", round(ganho, 2))
    return ganho
```
